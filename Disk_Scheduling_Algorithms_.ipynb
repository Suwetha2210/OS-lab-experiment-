```python
%%writefile disk.c
#include <stdio.h>
#include <stdlib.h>

#define MAX 50

/* ---------- FCFS ---------- */

void fcfs(int req[], int n, int head)
{
    int i;
    int seek = 0;

    printf("\n===== FCFS =====\n");
    printf("Head Movement: %d", head);

    for(i = 0; i < n; i++)
    {
        seek += abs(req[i] - head);
        head = req[i];
        printf(" -> %d", head);
    }

    printf("\nTotal Head Movement = %d\n", seek);
}


/* ---------- SSTF ---------- */

void sstf(int req[], int n, int head)
{
    int visited[MAX] = {0};
    int i, count;
    int index, min, distance;
    int seek = 0;

    printf("\n===== SSTF =====\n");
    printf("Head Movement: %d", head);

    for(count = 0; count < n; count++)
    {
        min = 99999;
        index = -1;

        for(i = 0; i < n; i++)
        {
            if(!visited[i])
            {
                distance = abs(req[i] - head);

                if(distance < min)
                {
                    min = distance;
                    index = i;
                }
            }
        }

        visited[index] = 1;
        seek += abs(req[index] - head);
        head = req[index];

        printf(" -> %d", head);
    }

    printf("\nTotal Head Movement = %d\n", seek);
}


/* ---------- SCAN ---------- */

void scan(int req[], int n, int head, int diskSize)
{
    int temp[MAX + 1];
    int i, j, count = 0;
    int seek = 0;
    int current;

    /* Copy requests */
    for(i = 0; i < n; i++)
        temp[i] = req[i];

    /* Add disk end */
    temp[n] = diskSize - 1;

    /* Sort */
    for(i = 0; i <= n; i++)
    {
        for(j = i + 1; j <= n; j++)
        {
            if(temp[i] > temp[j])
            {
                int x = temp[i];
                temp[i] = temp[j];
                temp[j] = x;
            }
        }
    }

    printf("\n===== SCAN =====\n");
    printf("Direction: Right\n");
    printf("Head Movement: %d", head);

    /* Move right */
    for(i = 0; i <= n; i++)
    {
        if(temp[i] >= head)
        {
            seek += abs(temp[i] - head);
            head = temp[i];

            printf(" -> %d", head);
            count++;
        }
    }

    /* Move left */
    for(i = n; i >= 0; i--)
    {
        if(temp[i] < head)
        {
            seek += abs(temp[i] - head);
            head = temp[i];

            printf(" -> %d", head);
        }
    }

    printf("\nTotal Head Movement = %d\n", seek);
}


/* ---------- C-SCAN ---------- */

void cscan(int req[], int n, int head, int diskSize)
{
    int temp[MAX + 2];
    int i, j;
    int seek = 0;

    /* Copy requests */
    for(i = 0; i < n; i++)
        temp[i] = req[i];

    /* Add disk boundaries */
    temp[n] = 0;
    temp[n + 1] = diskSize - 1;

    /* Sort */
    for(i = 0; i < n + 2; i++)
    {
        for(j = i + 1; j < n + 2; j++)
        {
            if(temp[i] > temp[j])
            {
                int x = temp[i];
                temp[i] = temp[j];
                temp[j] = x;
            }
        }
    }

    printf("\n===== C-SCAN =====\n");
    printf("Direction: Right\n");
    printf("Head Movement: %d", head);

    /* Move right */
    for(i = 0; i < n + 2; i++)
    {
        if(temp[i] >= head)
        {
            seek += abs(temp[i] - head);
            head = temp[i];

            printf(" -> %d", head);
        }
    }

    /* Jump to beginning */
    if(head != 0)
    {
        seek += diskSize - 1;
        head = 0;

        printf(" -> %d", head);
    }

    /* Move right again */
    for(i = 0; i < n + 2; i++)
    {
        if(temp[i] > 0)
        {
            seek += temp[i];
            head = temp[i];

            printf(" -> %d", head);
            break;
        }
    }

    printf("\nTotal Head Movement = %d\n", seek);
}


/* ---------- MAIN ---------- */

int main()
{
    int req[MAX];
    int n, head, diskSize;
    int i;

    printf("=====================================\n");
    printf("       DISK SCHEDULING ALGORITHMS\n");
    printf("=====================================\n");

    printf("\nEnter Number of Requests: ");
    scanf("%d", &n);

    if(n <= 0 || n > MAX)
    {
        printf("Invalid number of requests!\n");
        return 0;
    }

    printf("Enter Request Queue:\n");

    for(i = 0; i < n; i++)
        scanf("%d", &req[i]);

    printf("Enter Initial Head Position: ");
    scanf("%d", &head);

    printf("Enter Disk Size: ");
    scanf("%d", &diskSize);

    fcfs(req, n, head);
    sstf(req, n, head);
    scan(req, n, head, diskSize);
    cscan(req, n, head, diskSize);

    return 0;
}
```
