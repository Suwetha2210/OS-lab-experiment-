```python
%%writefile fileorg.c
#include <stdio.h>
#include <stdlib.h>

struct Student
{
    int regno;
    char name[30];
};

/* ---------- SEQUENTIAL ---------- */

void sequential()
{
    FILE *fp;
    struct Student s;

    printf("\n===== SEQUENTIAL FILE ORGANIZATION =====\n");

    fp = fopen("student.dat", "w");

    if(fp == NULL)
    {
        printf("Error creating file!\n");
        return;
    }

    printf("Enter Register Number: ");
    scanf("%d", &s.regno);

    printf("Enter Name: ");
    scanf("%29s", s.name);

    fprintf(fp, "%d %s\n", s.regno, s.name);
    fclose(fp);

    fp = fopen("student.dat", "r");

    if(fp == NULL)
    {
        printf("Error opening file!\n");
        return;
    }

    fscanf(fp, "%d %29s", &s.regno, s.name);

    printf("\nRecord Details\n");
    printf("Register Number : %d\n", s.regno);
    printf("Name            : %s\n", s.name);

    fclose(fp);
}

/* ---------- DIRECT / RANDOM ---------- */

void direct()
{
    struct Student s[3];
    int i, key;

    printf("\n===== DIRECT (RANDOM) FILE ORGANIZATION =====\n");

    for(i = 0; i < 3; i++)
    {
        printf("Enter Register Number for Student %d: ", i + 1);
        scanf("%d", &s[i].regno);

        printf("Enter Name: ");
        scanf("%29s", s[i].name);
    }

    printf("\nEnter Register Number to Search: ");
    scanf("%d", &key);

    for(i = 0; i < 3; i++)
    {
        if(s[i].regno == key)
        {
            printf("\nRecord Found\n");
            printf("Reg No : %d\n", s[i].regno);
            printf("Name   : %s\n", s[i].name);
            return;
        }
    }

    printf("\nRecord Not Found\n");
}

/* ---------- INDEXED ---------- */

void indexed()
{
    struct Student s[3];
    int i, key;

    printf("\n===== INDEXED FILE ORGANIZATION =====\n");

    printf("Enter 3 Student Records:\n");

    for(i = 0; i < 3; i++)
    {
        printf("\nStudent %d\n", i + 1);

        printf("Register Number: ");
        scanf("%d", &s[i].regno);

        printf("Name: ");
        scanf("%29s", s[i].name);
    }

    printf("\nEnter Register Number to Search: ");
    scanf("%d", &key);

    for(i = 0; i < 3; i++)
    {
        if(s[i].regno == key)
        {
            printf("\nRecord Found\n");
            printf("Reg No : %d\n", s[i].regno);
            printf("Name   : %s\n", s[i].name);
            return;
        }
    }

    printf("\nRecord Not Found\n");
}

/* ---------- MAIN ---------- */

int main()
{
    int choice;

    printf("=====================================\n");
    printf("      FILE ORGANIZATION METHODS\n");
    printf("=====================================\n");

    printf("\n1. Sequential File Organization");
    printf("\n2. Direct File Organization");
    printf("\n3. Indexed File Organization");
    printf("\n4. Exit");

    printf("\n\nEnter your choice: ");
    scanf("%d", &choice);

    switch(choice)
    {
        case 1:
            sequential();
            break;

        case 2:
            direct();
            break;

        case 3:
            indexed();
            break;

        case 4:
            printf("Exiting...\n");
            break;

        default:
            printf("Invalid Choice!\n");
    }

    return 0;
}
```
