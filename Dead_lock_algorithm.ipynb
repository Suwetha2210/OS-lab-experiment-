```python
%%writefile deadlock.c
#include <stdio.h>

int main()
{
    int allocation[10][10], request[10][10];
    int available[10], work[10], finish[10];
    int n, m;
    int i, j, k, found;
    int deadlock = 0;

    printf("Enter Number of Processes: ");
    scanf("%d", &n);

    printf("Enter Number of Resource Types: ");
    scanf("%d", &m);

    printf("\nEnter Allocation Matrix:\n");

    for(i = 0; i < n; i++)
    {
        for(j = 0; j < m; j++)
        {
            scanf("%d", &allocation[i][j]);
        }
    }

    printf("\nEnter Request Matrix:\n");

    for(i = 0; i < n; i++)
    {
        for(j = 0; j < m; j++)
        {
            scanf("%d", &request[i][j]);
        }
    }

    printf("\nEnter Available Resources:\n");

    for(i = 0; i < m; i++)
    {
        scanf("%d", &available[i]);
        work[i] = available[i];
    }

    /* Initialize Finish */
    for(i = 0; i < n; i++)
    {
        finish[i] = 0;
    }

    /* Deadlock Detection Algorithm */
    do
    {
        found = 0;

        for(i = 0; i < n; i++)
        {
            if(finish[i] == 0)
            {
                for(j = 0; j < m; j++)
                {
                    if(request[i][j] > work[j])
                    {
                        break;
                    }
                }

                if(j == m)
                {
                    for(k = 0; k < m; k++)
                    {
                        work[k] += allocation[i][k];
                    }

                    finish[i] = 1;
                    found = 1;
                }
            }
        }

    } while(found == 1);

    /* Check for deadlock */
    for(i = 0; i < n; i++)
    {
        if(finish[i] == 0)
        {
            deadlock = 1;
            break;
        }
    }

    if(deadlock == 0)
    {
        printf("\nNo Deadlock Detected\n");
    }
    else
    {
        printf("\nDeadlock Detected\n");
        printf("Deadlocked Processes: ");

        for(i = 0; i < n; i++)
        {
            if(finish[i] == 0)
            {
                printf("P%d ", i);
            }
        }

        printf("\n");
    }

    return 0;
}
```
