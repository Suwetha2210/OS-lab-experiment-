```python
%%writefile pagereplace.c
#include <stdio.h>
#include <limits.h>

#define MAX 50

/* ---------------- FIFO ---------------- */

void fifo(int pages[], int n, int frames)
{
    int memory[MAX];
    int i, j, k = 0;
    int fault = 0, found;

    for(i = 0; i < frames; i++)
        memory[i] = -1;

    for(i = 0; i < n; i++)
    {
        found = 0;

        for(j = 0; j < frames; j++)
        {
            if(memory[j] == pages[i])
            {
                found = 1;
                break;
            }
        }

        if(!found)
        {
            memory[k] = pages[i];
            k = (k + 1) % frames;
            fault++;
        }
    }

    printf("\nFIFO Page Faults = %d\n", fault);
}


/* ---------------- LRU ---------------- */

void lru(int pages[], int n, int frames)
{
    int memory[MAX], recent[MAX];
    int i, j, pos, found;
    int fault = 0;

    for(i = 0; i < frames; i++)
    {
        memory[i] = -1;
        recent[i] = 0;
    }

    for(i = 0; i < n; i++)
    {
        found = 0;

        for(j = 0; j < frames; j++)
        {
            if(memory[j] == pages[i])
            {
                found = 1;
                recent[j] = i;
                break;
            }
        }

        if(!found)
        {
            fault++;

            pos = 0;

            for(j = 1; j < frames; j++)
            {
                if(recent[j] < recent[pos])
                    pos = j;
            }

            memory[pos] = pages[i];
            recent[pos] = i;
        }
    }

    printf("LRU Page Faults  = %d\n", fault);
}


/* ---------------- OPTIMAL ---------------- */

void optimal(int pages[], int n, int frames)
{
    int memory[MAX];
    int i, j, k;
    int found, pos, farthest, next;
    int fault = 0;

    for(i = 0; i < frames; i++)
        memory[i] = -1;

    for(i = 0; i < n; i++)
    {
        found = 0;

        for(j = 0; j < frames; j++)
        {
            if(memory[j] == pages[i])
            {
                found = 1;
                break;
            }
        }

        if(found)
            continue;

        fault++;

        /* Find an empty frame */
        for(j = 0; j < frames; j++)
        {
            if(memory[j] == -1)
            {
                pos = j;
                break;
            }
        }

        /* If no empty frame, find optimal replacement */
        if(j == frames)
        {
            farthest = -1;
            pos = -1;

            for(j = 0; j < frames; j++)
            {
                next = -1;

                for(k = i + 1; k < n; k++)
                {
                    if(memory[j] == pages[k])
                    {
                        next = k;
                        break;
                    }
                }

                if(next == -1)
                {
                    pos = j;
                    break;
                }

                if(next > farthest)
                {
                    farthest = next;
                    pos = j;
                }
            }
        }

        memory[pos] = pages[i];
    }

    printf("Optimal Page Faults = %d\n", fault);
}


/* ---------------- MAIN ---------------- */

int main()
{
    int pages[MAX];
    int n, frames;
    int i;

    printf("Enter Number of Pages: ");
    scanf("%d", &n);

    printf("Enter Reference String:\n");

    for(i = 0; i < n; i++)
        scanf("%d", &pages[i]);

    printf("Enter Number of Frames: ");
    scanf("%d", &frames);

    if(frames <= 0 || frames > n)
    {
        printf("Invalid number of frames.\n");
        return 0;
    }

    printf("\n========== PAGE REPLACEMENT ==========\n");

    fifo(pages, n, frames);
    lru(pages, n, frames);
    optimal(pages, n, frames);

    return 0;
}
```
