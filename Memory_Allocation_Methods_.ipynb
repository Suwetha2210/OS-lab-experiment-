```python
%%writefile memory.c
#include <stdio.h>

#define MAX 20

void firstFit(int blocks[], int nb, int processes[], int np)
{
    int allocation[MAX];
    int i, j;

    for(i = 0; i < np; i++)
        allocation[i] = -1;

    for(i = 0; i < np; i++)
    {
        for(j = 0; j < nb; j++)
        {
            if(blocks[j] >= processes[i])
            {
                allocation[i] = j;
                blocks[j] -= processes[i];
                break;
            }
        }
    }

    printf("\n--- FIRST FIT ---\n");
    printf("Process\tSize\tBlock\n");

    for(i = 0; i < np; i++)
    {
        printf("P%d\t%d\t", i + 1, processes[i]);

        if(allocation[i] != -1)
            printf("B%d\n", allocation[i] + 1);
        else
            printf("Not Allocated\n");
    }
}

void bestFit(int blocks[], int nb, int processes[], int np)
{
    int allocation[MAX];
    int i, j, best;

    for(i = 0; i < np; i++)
        allocation[i] = -1;

    for(i = 0; i < np; i++)
    {
        best = -1;

        for(j = 0; j < nb; j++)
        {
            if(blocks[j] >= processes[i])
            {
                if(best == -1 || blocks[j] < blocks[best])
                    best = j;
            }
        }

        if(best != -1)
        {
            allocation[i] = best;
            blocks[best] -= processes[i];
        }
    }

    printf("\n--- BEST FIT ---\n");
    printf("Process\tSize\tBlock\n");

    for(i = 0; i < np; i++)
    {
        printf("P%d\t%d\t", i + 1, processes[i]);

        if(allocation[i] != -1)
            printf("B%d\n", allocation[i] + 1);
        else
            printf("Not Allocated\n");
    }
}

void worstFit(int blocks[], int nb, int processes[], int np)
{
    int allocation[MAX];
    int i, j, worst;

    for(i = 0; i < np; i++)
        allocation[i] = -1;

    for(i = 0; i < np; i++)
    {
        worst = -1;

        for(j = 0; j < nb; j++)
        {
            if(blocks[j] >= processes[i])
            {
                if(worst == -1 || blocks[j] > blocks[worst])
                    worst = j;
            }
        }

        if(worst != -1)
        {
            allocation[i] = worst;
            blocks[worst] -= processes[i];
        }
    }

    printf("\n--- WORST FIT ---\n");
    printf("Process\tSize\tBlock\n");

    for(i = 0; i < np; i++)
    {
        printf("P%d\t%d\t", i + 1, processes[i]);

        if(allocation[i] != -1)
            printf("B%d\n", allocation[i] + 1);
        else
            printf("Not Allocated\n");
    }
}

int main()
{
    int blocks[MAX], processes[MAX];
    int first[MAX], best[MAX], worst[MAX];
    int nb, np, i;

    printf("Enter Number of Blocks: ");
    scanf("%d", &nb);

    printf("Enter Number of Processes: ");
    scanf("%d", &np);

    printf("\nEnter Block Sizes:\n");

    for(i = 0; i < nb; i++)
        scanf("%d", &blocks[i]);

    printf("\nEnter Process Sizes:\n");

    for(i = 0; i < np; i++)
        scanf("%d", &processes[i]);

    /* Make copies for each algorithm */
    for(i = 0; i < nb; i++)
    {
        first[i] = blocks[i];
        best[i] = blocks[i];
        worst[i] = blocks[i];
    }

    firstFit(first, nb, processes, np);
    bestFit(best, nb, processes, np);
    worstFit(worst, nb, processes, np);

    return 0;
}
```
