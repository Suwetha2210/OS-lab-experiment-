```python
%%writefile semaphore_example.cpp
#include <iostream>
#include <unistd.h>
#include <semaphore.h>
#include <sys/mman.h>
#include <sys/wait.h>
#include <cstdlib>

using namespace std;

int main() {
    sem_t *sem;

    // Create shared memory for semaphore
    sem = (sem_t*)mmap(NULL, sizeof(sem_t),
                       PROT_READ | PROT_WRITE,
                       MAP_SHARED | MAP_ANONYMOUS,
                       -1, 0);

    // Initialize semaphore with value 1
    sem_init(sem, 1, 1);

    pid_t pid = fork();

    if (pid == 0) {
        // Child process
        sem_wait(sem);

        cout << "Child Process Entering Critical Section" << endl;
        sleep(2);
        cout << "Child Process Leaving Critical Section" << endl;

        sem_post(sem);
        exit(0);
    }
    else {
        // Parent process
        sem_wait(sem);

        cout << "Parent Process Entering Critical Section" << endl;
        sleep(2);
        cout << "Parent Process Leaving Critical Section" << endl;

        sem_post(sem);

        wait(NULL);
    }

    sem_destroy(sem);
    munmap(sem, sizeof(sem_t));

    return 0;
}
```

```python
%%bash
g++ -o semaphore_example semaphore_example.cpp
./semaphore_example
```
